# Pandas Indexing & Selection Reference

### 1. DataFrame Structure & Metadata
* `df.columns` — Returns an Index object containing all column headers.
* `df.index` — Returns the row index range or labels (`RangeIndex`).

---

### 2. Column Selection
* `df["Country"]` — **Single-column indexing:** Selects one column as a 1D **Series**.
* `df[["Country"]]` — **DataFrame indexing:** Passing a single-element list preserves the 2D **DataFrame** structure.
* `df[["Country", "Product"]]` — **Multi-column indexing:** Extracts multiple columns as a new **DataFrame**.
* `set(df["Country"])` — Extracts unique, unordered column values using Python's `set()`.
* `len(df["Country"])` — Returns the total count of rows/entries in the column.

---

### 3. Row Selection (`iloc`)
* `df.iloc[0]` — Selects a single row by 0-based integer position as a **Series**.
* `list(df.iloc[0])` — Converts row values into a standard Python `list`.
* `df.iloc[[0, 1, 2]]` — Selects multiple specific rows by positional indices as a **DataFrame**.

---

### 4. Element Access (Row & Column)
* `df.iloc[0, 8]` — **Positional access:** Selects scalar at row index `0`, column index `8`.
* `df.loc[0, "Country"]` — **Label-based access:** Selects scalar at row label `0`, column name `"Country"`.

---

### 5. Multi-Row & Multi-Column Slicing
* `df.iloc[[0, 1, 2], [1, 3]]` — Selects specific rows (`0, 1, 2`) and columns (`1, 3`) by **integer position**.
* `df.loc[[0, 1, 2], ["CustomerName", "Category"]]` — Selects specific rows (`0, 1, 2`) and columns (`"CustomerName", "Category"`) by **label**.

```python
print(df.columns)       # Column names
print(df.index)         # Row indices
```

```python
print(df["Country"]) # This will return a series, A 1D labeled array (like a column)
print(set(df["Country"])) # Set only keeps unique unordered elements
print(len(df["Country"])) # Prints the length of the column
print(len(set(df["Country"]))) # Prints the length of the set
```

```python
print(df[["Country", "Product"]]) # List-based column selection, passes a list of multiple column labels to [], extracting multiple columns as a DataFrame.
print(df[["Country"]]) # This is not a series, this is still a DataFrame
```

```python
print(df.iloc[0])        # iloc = integer-location; selects the first row as a Series (0-based positional index)
print(list(df.iloc[0]))  # Converts that row's values into a standard Python list
```

```python
print(df.iloc[[0,1,2]]) # List of specific rows
```

```python
print(df.iloc[0, 8])         # Positional access: row index 0, column index 8 (both integer-based)
print(df.loc[0, "Country"])  # Label-based access: row index label 0, column name "Country"
```

```python
df.iloc[[0, 1, 2], [1, 3]]           # Specific rows and specific columns by integer position
df.loc[[0, 1, 2], ["CustomerName", "Category"]]   # Specific rows and specific columns by label
```

# Boolean Indexing & Filtering Reference

### 1. Basic Equality & Comparison
* `df[df["Category"] == "Electronics"]` — **Equality filter:** Returns rows where the column matches an exact value.
* `df[df["Quantity"] > 5]` — **Comparison filter:** Returns rows where a numeric column meets a threshold (`>`, `<`, `>=`, `<=`, `!=`).

---

### 2. Multi-Condition Filtering (Bitwise Operators)
* `df[(df["Category"] == "Electronics") & (df["Country"] == "USA")]` — **Logical AND (`&`):** Both conditions must be True. Parentheses `(...)` around each condition are required.
* `df[(df["Category"] == "Electronics") | (df["Country"] == "USA")]` — **Logical OR (`|`):** At least one condition must be True. Parentheses `(...)` around each condition are required.

---

### 3. String Methods (`.str` Accessor)
* `df[df["CustomerName"].str.startswith("A")]` — Filters rows where the text starts with a given substring.
* `df[df["CustomerName"].str.endswith("n")]` — Filters rows where the text ends with a given substring.
* `df[df["CustomerName"].str.contains("text", case=False)]` — Filters rows where the text contains a substring (case-insensitive flag optional).

---

### 4. Membership & Inversion (`.isin()` & `~`)
* `df[df["Country"].isin(["USA", "Sweden", "Brazil"])]` — **Inclusion:** Matches values present in an iterable list.
* `df[~df["Country"].isin(["USA", "Sweden", "Brazil"])]` — **Inversion/NOT (`~`):** Negates the boolean Series to return values NOT in the list.

```python
# Boolean indexing: filters and returns only the rows where the "Category" column equals "Electronics"
df[df["Category"] == "Electronics"]
```

```python
# Multi-condition boolean filtering: returns rows where both conditions are True using element-wise AND (&); parentheses are required around each condition
df[(df["Category"] == "Electronics") & (df["Country"] == "USA")]
```

```python
# Multi-condition boolean filtering: returns rows where at least one condition is True using element-wise OR (|); parentheses are required around each condition
df[(df["Category"] == "Electronics") | (df["Country"] == "USA")]
```

```python
# Comparison-based filtering: returns rows where the numeric value in "Quantity" is strictly greater than 5
df[df["Quantity"] > 5]
```

```python
# String filtering: returns rows where "CustomerName" starts with the letter "R" using the vectorized string accessor (.str)
df[df["CustomerName"].str.startswith("R")]
```

```python
# String filtering: returns rows where "CustomerName" ends with the lowercase letter "n" using the vectorized string accessor (.str)
df[df["CustomerName"].str.endswith("n")]
```

```python
# Filters rows containing "john" anywhere in the text, ignoring case and skipping NaN values safely
df[df["CustomerName"].str.contains("john", case=False, na=False)]
# case=False: Ignores capitalization. Searching "john" will match "John", "JOHN", or "john". Default is case=True
# na=False: Crucial if your column has missing/null values (NaN). Without setting na=False, missing entries return NaN instead of False, which can raise an error when filtering
```

```python
# Membership filtering: returns rows where "Country" matches any value present in the specified list using .isin()
df[df["Country"].isin(["USA", "Sweden", "Brazil"])]
```

```python
# Inverted membership filtering: the tilde (~) negates the condition, returning rows where "Country" is NOT in the specified list
df[~df["Country"].isin(["USA", "Sweden", "Brazil"])]
```

# Modifying & Updating DataFrames Reference

### 1. Conditional Value Updating (`.loc`)
The syntax `df.loc[row_condition, "ColumnName"] = new_value` updates specific fields in-place without triggering chained assignment warnings:
* `df.loc[df["CustomerName"] == "Anna Ivanova", "Product"] = "Desk Pad"`  
  Finds matching rows and updates the target column value (e.g., changes `"Desk Clock"` to `"Desk Pad"`).
* `df.loc[df["Country"] == "USA", "Country"] = "United States"`  
  Replaces all matching entries across the column in-place.

---

### 2. Vectorized Column Transformation
* `df["Country"] = df["Country"].str.upper()`  
  Transforms all text values in the column to uppercase and overwrites the column in-place.

```python
# Label-based boolean indexing: returns rows where "CustomerName" equals "Anna Ivanova" using .loc
df.loc[df["CustomerName"] == "Anna Ivanova"]
```

```python
# Conditional column update: changes "Product" from "Desk Clock" to "Desk Pad" where "CustomerName" is "Anna Ivanova"
df.loc[df["CustomerName"] == "Anna Ivanova", "Product"] = "Desk Pad"

# Filters and displays the updated row(s) for "Anna Ivanova" to verify the change
print(df.loc[df["CustomerName"] == "Anna Ivanova"])
```

```python
# In-place value replacement: renames all occurrences of "USA" to "United States" in the "Country" column
df.loc[df["Country"] == "USA", "Country"] = "United States"

# Filters and displays all rows where "Country" is now "United States" to confirm the replacement
print(df.loc[df["Country"] == "United States"])
```

```python
df["Country"] = df["Country"].str.upper()
print(df["Country"])
```

# DataFrame Cleaning & Structural Modification Reference

### 1. Dropping Rows
* `df = df.drop(39)`  
  Removes the row with index label `39` and reassigns the resulting DataFrame.

---

### 2. Handling Missing Data (`NaN`)
* `df = df.dropna()`  
  Drops any row containing at least one missing value across columns and reassigns the cleaned DataFrame.
* `df.fillna({"OrderID": 0}, inplace=True)`  
  Replaces missing values in the `"OrderID"` column with `0` directly in memory without requiring variable reassignment.

---

### 3. Renaming Columns
* `df.rename(columns={"OrderID": "Order ID"}, inplace=True)`  
  Renames specified column labels using a dictionary map directly in memory.

```python
# Row deletion: removes the row with index label 39 and reassigns the updated DataFrame
df = df.drop(39)

# Displays the last 5 rows to confirm index 39 is removed
print(df.tail())
```

```python
# Drops any row containing at least one NaN and reassigns the cleaned result back to df
df = df.dropna()

# Replaces NaN values in "OrderID" with 0 directly in memory without needing reassignment (returns None)
df.fillna({"OrderID": 0}, inplace=True)
```

```python
# Renames the column "OrderID" to "Order ID" directly in memory without needing reassignment
df.rename(columns={"OrderID": "Order ID"}, inplace=True)
print(df)
```

# Sorting & Export Reference

### 1. In-Place Sorting
* `df.sort_values("Price", ascending=False, inplace=True)`  
  Sorts the DataFrame by `"Price"` in descending order (highest to lowest) directly in memory without reassignment.

---

### 2. Exporting to CSV
* `df.to_csv("orders_modified.csv", index=False)`  
  Exports the DataFrame to a CSV file named `"orders_modified.csv"`, omitting the default integer row index to prevent extra unnamed columns upon re-reading.

```python
# Sorting rows: sorts the DataFrame by "Price" in descending order (highest to lowest)
df.sort_values("Price", ascending=False, inplace=True)
```

```python
# File export: saves the DataFrame to a CSV file named "orders_modified.csv" without writing row index numbers
df.to_csv("orders_modified.csv", index = False)
```
